### 1. Queue using Stack

```python
class QueueUsingStack:
    def __init__(self):
        self.s1 = deque()
        self.s2 = deque()
    

    def enqueue(self, data):
        self.s1.append(data)
    

    def dequeue(self):
        if not self.s2:
            while self.s1:
                self.s2.append(self.s1.pop())
        if not self.s2:
            print('Queue is empty.. cannot Dequeue')
            return
        return self.s2.pop()
```

```python
my_queue = QueueUsingStack()
my_queue
```

```python
my_queue.dequeue()
```

```python
my_queue.enqueue(10)
my_queue.enqueue(20)
my_queue.enqueue(30)
```

### 2. Reverse first `k` in a Queue

- `q`: given queue

```python
# input1: k = 3
q1 = deque()
for x in range(1, 6):
    q1.append(x)

# input2: k = 4
q2 = deque()
for x in range(4, 0, -1):
    q2.append(x)

print(q1, q2)
```

```python
def reverse_first_k(q, k):
    stack = deque()
    result = deque()

    for _ in range(k):
        stack.append(q.popleft())
    
    while stack:
        result.append(stack.pop())
    while q:
        result.append(q.popleft())
    return result
```

```python
reverse_first_k(q1, 3)
```
