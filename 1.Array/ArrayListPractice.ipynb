# 1. Find the Largest Number in a List

```python
numbers = [12, 45, 7, 89, 34, 95, 62, 57]

largest = numbers[0]
for i in numbers:
    if i > largest:
        largest=i
        
print(largest)
```

# 2. Count Occurrences of an Element

```python
items = ["apple", "banana", "apple", "orange", "banana", "apple"]
target = "banana"

count = 0
for i in items:
    if i == target:
        count=count+1

print(f'{target} appears {count} times')
```

# 3. Remove Duplicates from a List

```python
numbers = [1, 2, 2, 3, 4, 4, 5]

uni_ele =[]
for i in numbers:
    if i not in uni_ele:
        uni_ele.append(i)

print(f'Unique elements:', uni_ele)
```

# 4. Reverse a List (without the reverse method)

```python
data = [4, 6, 2, 3, 1, 7, 5]

n = len(data)
rev_li=[]
for i in range(n-1, -1, -1):
    rev_li.append(data[i])

print(f'The reversed data is:', rev_li)
```

```python
n = len(data) - 1
reversed_data2 = []

while n >= 0:
    reversed_data2.append(data[n])
    n -= 1
print(f'The reversed data is:', reversed_data2)
```
